```python
storage_account_name = ""
storage_account_key = ""

spark.conf.set(
    f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net",
    storage_account_key
)
```

```python
silver_path = f"abfss://processed@{storage_account_name}.dfs.core.windows.net/911-recordings/features_silver"

silver_df = spark.read.parquet(silver_path)

print("=== SILVER DATA LOADED ===")
print(f"Rows    : {silver_df.count()}")
print(f"Columns : {len(silver_df.columns)}")
display(silver_df.limit(5))
```

```python
import pandas as pd
import numpy as np

# Define feature columns only (exclude metadata and label columns)
meta_cols    = ["file_name", "label", "label_encoded", "civilian_initiated", 
                "deaths", "potential_death", "false_alarm"]
feature_cols = [c for c in silver_df.columns if c not in meta_cols]

print(f"Features before variance filter: {len(feature_cols)}")

# Convert features to pandas for variance calculation
features_pd = silver_df.select(feature_cols).toPandas()

# Calculate variance for each feature
variances = features_pd.var()

# Keep features with variance above threshold
VARIANCE_THRESHOLD = 0.01
low_variance = variances[variances < VARIANCE_THRESHOLD].index.tolist()
high_variance = variances[variances >= VARIANCE_THRESHOLD].index.tolist()

print(f"Low variance features removed : {len(low_variance)}")
print(f"Features remaining            : {len(high_variance)}")

if low_variance:
    print(f"Removed: {low_variance}")

# Keep only high variance features + metadata cols
cols_to_keep = meta_cols + high_variance
silver_df_filtered = silver_df.select(cols_to_keep)

print(f"\nDataFrame after variance filter:")
print(f"Rows    : {silver_df_filtered.count()}")
print(f"Columns : {len(silver_df_filtered.columns)}")
```

```python
# Calculate correlation matrix
corr_matrix = features_pd[high_variance].corr().abs()

# Upper triangle of correlation matrix
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Find features with correlation above threshold
CORRELATION_THRESHOLD = 0.95
highly_correlated = [
    col for col in upper_triangle.columns 
    if any(upper_triangle[col] > CORRELATION_THRESHOLD)
]

uncorrelated = [
    col for col in high_variance 
    if col not in highly_correlated
]

print(f"Features before correlation filter : {len(high_variance)}")
print(f"Highly correlated features removed : {len(highly_correlated)}")
print(f"Features remaining                 : {len(uncorrelated)}")

if highly_correlated:
    print(f"\nRemoved: {highly_correlated}")

# Keep only uncorrelated features + metadata cols
final_cols = meta_cols + uncorrelated
gold_df = silver_df_filtered.select(final_cols)

print(f"\nDataFrame after correlation filter:")
print(f"Rows    : {gold_df.count()}")
print(f"Columns : {len(gold_df.columns)}")
```

```python
from pyspark.sql.functions import col, when, count as spark_count

print("=== GOLD LAYER VALIDATION ===")
print(f"Total rows    : {gold_df.count()}")
print(f"Total columns : {len(gold_df.columns)}")

print("\n=== LABEL DISTRIBUTION ===")
gold_df.groupBy("label", "label_encoded").count().show()

print("\n=== NULL CHECK ===")
gold_df.select([
    spark_count(when(col(c).isNull(), c)).alias(c)
    for c in meta_cols
]).show()

print("\n=== FEATURE STATS SAMPLE ===")
gold_df.describe(uncorrelated[:5]).show()
```

```python
gold_path = f"abfss://curated@{storage_account_name}.dfs.core.windows.net/911-recordings/features_gold"

gold_df.write.mode("overwrite").parquet(gold_path)

print(f"Gold layer written successfully")
print(f"Path: {gold_path}")

# Read back and verify
gold_check = spark.read.parquet(gold_path)

print("\n=== GOLD LAYER VERIFICATION ===")
print(f"Rows    : {gold_check.count()}")
print(f"Columns : {len(gold_check.columns)}")
print(f"\nSchema:")
gold_check.printSchema()
display(gold_check.limit(5))
```
