```python
import pandas as pd

gold_path = f"abfss://curated@{storage_account_name}.dfs.core.windows.net/911-recordings/features_gold"

gold_df = spark.read.parquet(gold_path)

# Convert to pandas for visualization
df = gold_df.toPandas()

print(f"Rows    : {len(df)}")
print(f"Columns : {len(df.columns)}")
display(gold_df.limit(5))
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import numpy as np

# Style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"]     = 100

# Label mapping for display
label_names  = {0: "Medical", 1: "Fire", 2: "Violence"}
label_colors = {"Medical": "#2196F3", "Fire": "#F44336", "Violence": "#FF9800"}

df["label_name"] = df["label_encoded"].map(label_names)

print("Libraries loaded ✅")
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

counts = df["label_name"].value_counts()
colors = [label_colors[l] for l in counts.index]

bars = ax.bar(counts.index, counts.values, color=colors, edgecolor="white", linewidth=0.8)

# Add count labels on bars
for bar, count in zip(bars, counts.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 3,
        str(count),
        ha="center", va="bottom", fontsize=12, fontweight="bold"
    )

ax.set_title("Label Distribution — 911 Emergency Calls", fontsize=14, fontweight="bold")
ax.set_xlabel("Emergency Category", fontsize=12)
ax.set_ylabel("Number of Calls", fontsize=12)
ax.set_ylim(0, counts.max() + 40)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 7))

counts  = df["label_name"].value_counts()
colors  = [label_colors[l] for l in counts.index]
explode = [0.03] * len(counts)

wedges, texts, autotexts = ax.pie(
    counts.values,
    labels=counts.index,
    autopct="%1.1f%%",
    colors=colors,
    explode=explode,
    startangle=140,
    textprops={"fontsize": 12}
)

for autotext in autotexts:
    autotext.set_fontweight("bold")

ax.set_title("Class Distribution — Emergency Categories", fontsize=14, fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
mfcc_mean_cols = [f"mfcc_{i}_mean_scaled" for i in range(1, 14)]

mfcc_means = df.groupby("label_name")[mfcc_mean_cols].mean()

fig, ax = plt.subplots(figsize=(12, 5))

for label, color in label_colors.items():
    if label in mfcc_means.index:
        ax.plot(
            range(1, 14),
            mfcc_means.loc[label].values,
            marker="o",
            label=label,
            color=color,
            linewidth=2
        )

ax.set_title("Mean MFCC Coefficients per Emergency Category", fontsize=14, fontweight="bold")
ax.set_xlabel("MFCC Coefficient", fontsize=12)
ax.set_ylabel("Mean Scaled Value", fontsize=12)
ax.set_xticks(range(1, 14))
ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

for label, color in label_colors.items():
    subset = df[df["label_name"] == label]["rms_energy_mean_scaled"]
    subset.plot.kde(ax=ax, label=label, color=color, linewidth=2)

ax.set_title("RMS Energy Distribution per Emergency Category", fontsize=14, fontweight="bold")
ax.set_xlabel("RMS Energy (Scaled)", fontsize=12)
ax.set_ylabel("Density", fontsize=12)
ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
spectral_cols = [
    "spectral_centroid_mean_scaled",
    "spectral_rolloff_mean_scaled",
    "spectral_bandwidth_mean_scaled",
    "zcr_mean_scaled"
]

spectral_labels = [
    "Spectral Centroid",
    "Spectral Rolloff",
    "Spectral Bandwidth",
    "Zero Crossing Rate"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, (col, label) in enumerate(zip(spectral_cols, spectral_labels)):
    data_to_plot = [
        df[df["label_name"] == l][col].values
        for l in label_colors.keys()
    ]
    bp = axes[i].boxplot(
        data_to_plot,
        labels=list(label_colors.keys()),
        patch_artist=True,
        medianprops=dict(color="black", linewidth=2)
    )
    for patch, color in zip(bp["boxes"], label_colors.values()):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    axes[i].set_title(label, fontsize=12, fontweight="bold")
    axes[i].set_ylabel("Scaled Value", fontsize=10)

fig.suptitle("Spectral Features per Emergency Category", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
top_features = [
    "mfcc_1_mean_scaled",
    "mfcc_2_mean_scaled",
    "rms_energy_mean_scaled",
    "spectral_centroid_mean_scaled"
]

top_labels = [
    "MFCC 1 Mean",
    "MFCC 2 Mean",
    "RMS Energy Mean",
    "Spectral Centroid Mean"
]

fig, axes = plt.subplots(1, 4, figsize=(16, 6))

for i, (col, label) in enumerate(zip(top_features, top_labels)):
    sns.violinplot(
        data=df,
        x="label_name",
        y=col,
        ax=axes[i],
        palette=label_colors,
        inner="box",
        linewidth=1.2
    )
    axes[i].set_title(label, fontsize=11, fontweight="bold")
    axes[i].set_xlabel("")
    axes[i].set_ylabel("Scaled Value", fontsize=9)
    axes[i].tick_params(axis="x", rotation=15)

fig.suptitle("Class Separability — Top Acoustic Features", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```
